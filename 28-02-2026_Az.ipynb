### Categorical Data in Pandas

class_data = {
    "Name" : ["A","B","C","D"],
    "Groups: ["G1',"G2","G3,"G4"]
}

```python
class_data = {
    "Name": ["A", "B", "C", "D"],
    "Groups": ["G1", "G2", "G3", "G4"] # Fixed the quote here
}
```

```python
import pandas as pd
```

```python
class_df = pd.DataFrame(class_data)
```

```python
class_df.head()
```

```python
class_df["Group_cat"] = pd.Categorical(
    class_df["Groups"],
    categories = ["G3", "G1", "G2", "G4"],
    ordered = True
)
```

```python
class_df["Group_cat"]
```

```python
class_df["Group_cat"] >= "G4"
```

```python
titanic_dataset_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"

df = pd.read_csv(titanic_dataset_url)
```

```python
df.head()
```

```python
df["Embarked"].unique()
```

```python
df["Embarked"] = df["Embarked"].fillna("N")
```

```python
df["Embarked_Category"] = pd.Categorical(
    df["Embarked"],
    categories= ["N","S","C","Q"],
    ordered = True
)
```

```python
df["Embarked_Category"].cat.categories
```

```python
df.head()
```

```python
df["Sex"].unique()
```

```python
df["sex_custom"] = df["Sex"].astype("category").cat.codes
```

```python
df.head()
```

```python
df["sex_custom_cat"] = df["Sex"].apply(lambda sex:1 if sex == "male" else 0)
```

```python
df.head()
```

```python
df["Sex_replace_custom"] = df["Sex"] = df["Sex"].replace(['female', 'male'], [0, 1])
```

```python
df.head()
```

```python
df.info()
```

```python
# Raw data dictionary
data = {
    "Event": [
        "Sputnik 1 Launch", 
        "Vostok 1 (Yuri Gagarin)", 
        "Apollo 11 Moon Landing", 
        "Mars Pathfinder Landing", 
        "James Webb Telescope Launch"
    ],
    "Date": [
        "1957-10-04", 
        "1961-04-12", 
        "1969-07-20", 
        "1997-07-04", 
        "2021-12-25"
    ]
}

# Create the DataFrame
df_events = pd.DataFrame(data)

print(df_events)
df_events.info()
```

```python
# 1. IMPORTANT: Convert the 'Date' column to actual datetime objects
df_events["Date"] = pd.to_datetime(df_events["Date"])

# 2. Extract the Month Name
df_events["Month"] = df_events["Date"].dt.month_name()

# 3. Extract other useful info
df_events["Year"] = df_events["Date"].dt.year
df_events["Day_of_Week"] = df_events["Date"].dt.day_name()

print(df_events)

df_events.info()
```

```python
df_events["custom_date_category"] = df_events["Date"].astype("category").cat.codes
```
